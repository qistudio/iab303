### Moving from analyst centric to stakeholder centric language

|Phase         | Analyst             | Stakeholder             |
|:------------:|:-------------------:|:-----------------------:|
| Question     | Which product has highest number of sales? | Which products are the strongest performers and why? |
| Data         | Load in sales data and clean for NaNs      | Look at sales data for potential high performing products |
| Analysis     | Group by product and find Max              | Analyse in terms of both quantity of sales, revenue, and margin.Which products are high volume strong, and which are high margin strong?|
| Visualisation | Show a chart of product sales quantity    | Visualise the comparison between strong and week performers in terms of quantity of sales, total revenue, and margin |
| Insights    | Product X had the most sales                | Most revenue is coming from X and Y, although Z is high margin low volume  and therefore may have lower cost of sales. |

### Structuring reporting for executive stakeholders

Structure data analytics reports in a way that is predictable and easy for the reader to find the required information. For example:

* Executive summary
* Outline/contents
* Introduction 
    - business concern
    - stakeholders
    - questions
* Body
    - analysis
    - tables & visualisations
    - key facts & findings
    - ethical considerations
* Conclusion
    - recommendations
    - limitations
