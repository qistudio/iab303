## Structured Data - Strengths and Weaknesses

Amazon Movies revenue and subscribers by regions





```python
import pandas as pd

# Read and analyse structured data
df = pd.read_csv("data/DataAmazonMoviesRevenue2020.csv")
df["Quarter"] = df["Years"].apply(lambda x: x.split("-")[0])
df["Years"] = df["Years"].apply(lambda x: x.split("-")[1]).astype("int")
years_df = df[["Area", "Years", "Revenue","Subscribers"]].groupby(["Years", "Area"]).sum()
years_df
```

```python
df
```

```python
import plotly.express as px

# Convert grouped data to dataframe
years_df = years_df.reset_index()

# Visualise revenue
fig = px.bar(years_df,y='Revenue',x='Years',color='Area',barmode='group',title="Amazon Movie Revenue")
fig.update_xaxes(type='category')
fig.show()
```

```python
# Visualise subscribers
fig = px.bar(years_df,y='Subscribers',x='Years',color='Area',barmode='group',title="Amazon Movie Subscribers")
fig.update_xaxes(type='category')
fig.show()
```

## Semi-structured data - threats and opportunities

Amazon Movies reviews were already analysed inweek 4 tutorial. Here's a recap...

```python
# Import python regular expression library
import re

# Read the file
with open('data/amazon-movies-reviews.txt', 'r') as file:
    text = file.read()

# Get reviews
reviews = text.split("\n\n")
reviews
```

We need to extract the relevant information from the semi-structured data

```python
# Create a new dataframe that will store the extracted data
reviews_df = pd.DataFrame(columns=(["ProductId", "Score", "Region", "Summary"]))

# Find that patterns to extract the productId, score and summary
for review_text in reviews:
    review = {} # empty dict for review
    for line in review_text.split("\n"):
        match_id = re.match(r"^product[-|/]product[I|i][D|d][\s:|:]", line)
        if (match_id):
            review["ProductId"] = re.split(r":\s", line)[1]
        match_score = re.match(r"^review[-|/]score[\s:|:]", line)
        if (match_score):
            review["Score"] = re.split(r":\s", line)[1]
        match_region = re.match(r"^review/region:", line)
        if (match_region):
            review["Region"] = re.split(r":\s", line)[1]
        match_summary = re.match(r"^review/summary:", line)
        if (match_summary):
            review["Summary"] = re.split(r":\s", line)[1]
    #print(review)
    new_df = pd.DataFrame([review])      
    reviews_df = pd.concat([reviews_df,new_df], ignore_index=True)

# Convert numeric value from text to float
reviews_df["Score"] = reviews_df["Score"].astype("float")
reviews_df
```

```python
# Group by movie and regions to see the scores
movie_region_df = reviews_df[["ProductId", "Region", "Score"]].groupby(["ProductId", "Region"]).mean()
movie_region_df
```

```python
# Convert grouped data to dataframe
movie_sorted_df = movie_region_df.sort_values(by="Score", ascending=False).reset_index()

# Visualise 
fig = px.bar(movie_sorted_df,y='Score',x='ProductId',color='Region',barmode='group')
fig.update_xaxes(type='category')
fig.show()
```
